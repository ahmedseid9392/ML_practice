```python
import numpy as np

scores = np.array([70, 80, 90])
print(np.mean(scores))  # 80.0
```

```python
import numpy as np

scores = np.array([70, 80, 90, 85])

students = np.array([
    [70, 80, 90],
    [60, 75, 85]
])

print(scores.shape)
print(students.shape)

print(scores.ndim)
print(students.ndim)
```

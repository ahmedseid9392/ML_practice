```python

import pandas as pd
import matplotlib.pyplot as plt

# 1. Create the dataset
data = {
    "study_hours": [1, 2, 3, 4, 5],
    "exam_score": [35, 45, 55, 70, 85]
}

df = pd.DataFrame(data)

# 2. Analyze the data
print("Dataset:")
print(df)

print("\nSummary statistics:")
print(df.describe())

print("\nMissing values:")
print(df.isnull().sum())

# 3. Visualize the relationship
plt.scatter(df["study_hours"], df["exam_score"])

plt.title("Study Hours vs Exam Scores")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Define data
x = np.array([[23000], [14000], [24000], [52500], [18000], [15000], [16000], [41500], [45000]])
y = np.array([9500, 5000, 10000, 18000, 6000, 5000, 6000, 18000, 17500])

# 2. Fit the Linear Regression model
model = LinearRegression()  
model.fit(x, y)

# 3. Calculate Key Analysis Metrics
slope = model.coef_[0]
intercept = model.intercept_
r_squared = model.score(x, y)
prediction = model.predict([[22000]])[0]

print("--- DATA ANALYSIS METRICS ---")
print(f"Slope (Coefficient): {slope:.4f}")
print(f"Intercept:           {intercept:.2f}")
print(f"R-squared (Fit):     {r_squared:.4f}")
print(f"Prediction for 22k:  {prediction:.2f}\n")

# 4. Representation & Data Visualization
plt.figure(figsize=(8, 5))

# Plot actual data points
plt.scatter(x, y, color='blue', label='Actual Data', zorder=5)

# Plot the regression line
x_line = np.linspace(x.min(), x.max(), 100).reshape(-1, 1)
y_line = model.predict(x_line)
plt.plot(x_line, y_line, color='red', linewidth=2, label=f'Regression Line (y = {slope:.2f}x + {intercept:.2f})')

# Highlight the specific prediction point
plt.scatter([[22000]], [prediction], color='green', marker='X', s=150, label=f'Prediction for 22k ({int(prediction)})', zorder=6)

# Formatting the plot
plt.title('Linear Regression Model & Data Analysis', fontsize=14, fontweight='bold')
plt.xlabel('X Values', fontsize=12)
plt.ylabel('Y Values', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

# Show the plot
plt.show()

```
